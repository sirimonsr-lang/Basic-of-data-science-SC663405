## 2. ชนิดข้อมูล (Data Types)

Python มีชนิดข้อมูลพื้นฐานหลายแบบ แต่ที่ใช้บ่อยที่สุด 4 แบบคือ:

| ชนิดข้อมูล | ชื่อเต็ม | ตัวอย่าง | ใช้เก็บอะไร |
|---|---|---|---|
| `int` | Integer | `10`, `-5`, `0` | เลขจำนวนเต็ม |
| `float` | Floating point | `3.14`, `-0.5` | เลขทศนิยม |
| `str` | String | `"สวัสดี"`, `'Hi'` | ข้อความ |
| `bool` | Boolean | `True`, `False` | ค่าจริง/เท็จ |

https://www.w3schools.com/python/python_datatypes.asp

ใช้ฟังก์ชัน `type()` เพื่อตรวจสอบชนิดข้อมูลของตัวแปรได้


````python
# ตรวจสอบชนิดข้อมูลด้วย type()
a = 10
b = 3.14
c = "Data Science"
d = False # True

print(type(a))   # int
print(type(b))   # float
print(type(c))   # str
print(type(d))   # bool
````

### 2.1 int (จำนวนเต็ม)

ใช้เก็บเลขจำนวนเต็ม ทั้งบวกและลบ ไม่มีจุดทศนิยม


````python
# ตัวอย่าง int
num_students = 45
temperature_celsius = -2
year = 2569

print(num_students, temperature_celsius, year)
print(type(num_students))
````

### 2.2 float (ทศนิยม)

ใช้เก็บเลขที่มีจุดทศนิยม — แม้จะใส่ `.0` ก็ถือเป็น float ทันที


````python
# ตัวอย่าง float
gpa = 3.75
pi_value = 3.14159
price = 99.0   # มี .0 ก็เป็น float แล้ว

print(gpa, pi_value, price)
print(type(price))
````

### 2.3 str (ข้อความ)

ใช้เก็บข้อความ ครอบด้วยเครื่องหมายคำพูดเดี่ยว `'...'` หรือคู่ `"..."` ก็ได้ — ผลลัพธ์เหมือนกัน


````python
# ตัวอย่าง str
greeting = "สวัสดีค่ะ"
course_code = 'SC 612 104'

print(greeting)
print(course_code)

# เชื่อมข้อความด้วย +
full_text = greeting + " " + course_code
print(full_text)

# str ความยาวหลายบรรทัด ใช้ '''...''' หรือ """..."""
message = """บรรทัดที่ 1
บรรทัดที่ 2"""
print(message)
````

## 3. Operators (ตัวดำเนินการ)


แบ่งเป็น 3 กลุ่มหลักที่ต้องรู้จัก: **คณิตศาสตร์**, **เปรียบเทียบ**, และ **ตรรกะ**

### 3.1 Arithmetic Operators (ตัวดำเนินการทางคณิตศาสตร์)

| Operator | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|---|---|---|---|
| `+` | บวก | `5 + 3` | `8` |
| `-` | ลบ | `5 - 3` | `2` |
| `*` | คูณ | `5 * 3` | `15` |
| `/` | หาร (ได้ float เสมอ) | `5 / 2` | `2.5` |
| `//` | หารปัดเศษทิ้ง (floor division) | `5 // 2` | `2` |
| `%` | หารเอาเศษ (modulo) | `5 % 2` | `1` |
| `**` | ยกกำลัง | `5 ** 2` | `25` |

````python
a = 17
b = 5

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)    # หารปกติ ได้ float
print("a // b =", a // b)   # หารปัดเศษทิ้ง
print("a % b  =", a % b)    # เศษจากการหาร
print("a ** b =", a ** b)   # a ยกกำลัง b
````

**จุดที่ต้องระวัง:** `/` กับ `//` ให้ผลต่างกัน!
- `/` คือหารแบบปกติ ผลลัพธ์เป็น `float` เสมอ แม้จะหารลงตัว
- `//` คือหารแบบปัดเศษทิ้ง (floor) ผลลัพธ์เป็น `int` ถ้า input เป็น int ทั้งคู่

`%` (modulo) มีประโยชน์มากในการเช็คว่าเลขคู่/เลขคี่ หรือเช็คตัวหารร่วม


````python
# ตัวอย่างการใช้ % เช็คเลขคู่/เลขคี่
number = 10

if number % 2 == 0:
    print(f"{number} เป็นเลขคู่")
else:
    print(f"{number} เป็นเลขคี่")
````

#### ลำดับการคำนวณ (Order of Operations / PEMDAS)

เมื่อมีตัวดำเนินการหลายตัวอยู่ในนิพจน์เดียว Python จะคำนวณตามลำดับความสำคัญ (Precedence)
เหมือนที่เราคุ้นเคยในคณิตศาสตร์ ซึ่งเรียกย่อว่า PEMDAS/BODMAS

- **P**arentheses (วงเล็บ `()`) — ทำในวงเล็บก่อนเสมอ
- **E**xponents (เลขยกกำลัง `**`) — ทำเป็นลำดับถัดมา
- **M**ultiplication (คูณ `*`) และ **D**ivision (หาร `/`, `//`, `%`) — มีความสำคัญเท่ากัน ทำจากซ้ายไปขวา
- **A**ddition (บวก `+`) และ **S**ubtraction (ลบ `-`) — มีความสำคัญเท่ากัน ทำจากซ้ายไปขวา

การใช้วงเล็บ `()` สามารถบังคับลำดับการคำนวณให้เป็นไปตามที่เราต้องการได้

--- ตัวอย่าง PEMDAS ---


````python
# ไม่มีวงเล็บ: คูณ/หาร ทำก่อน บวก/ลบ
# 10 + 5 * 2
result1 = 10 + 5 * 2
print(f"10 + 5 * 2 = {result1}")
````

````python
# มีวงเล็บ: ทำในวงเล็บก่อน
# (10 + 5) * 2
result2 = (10 + 5) * 2
print(f"(10 + 5) * 2 = {result2}")
````

````python
# ตัวอย่างที่ซับซ้อนขึ้น:
result3 = 2 ** 3 + 6 / 3 - 1
print(f"2 ** 3 + 6 / 3 - 1 = {result3}")
````

````python
# ใช้วงเล็บเพื่อเปลี่ยนลำดับ
result4 = 2 ** (3 + 6) / 3 - 1
print(f"2 ** (3 + 6) / 3 - 1 = {result4}")
````

````python
print("\nข้อควรจำ: เมื่อมี operator ความสำคัญเท่ากัน ให้ทำจากซ้ายไปขวา")
# 10 / 2 * 5
result5 = 10 / 2 * 5
print(f"10 / 2 * 5 = {result5}")
````

````python
# 10 * 5 / 2
result6 = 10 * 5 / 2
print(f"10 * 5 / 2 = {result6}")
````

### 3.2 Comparison Operators (ตัวดำเนินการเปรียบเทียบ)

ผลลัพธ์จากการเปรียบเทียบเป็น `bool` เสมอ (`True` หรือ `False`)

| Operator | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|---|---|---|---|
| `==` | เท่ากับ | `5 == 5` | `True` |
| `!=` | ไม่เท่ากับ | `5 != 3` | `True` |
| `>` | มากกว่า | `5 > 3` | `True` |
| `<` | น้อยกว่า | `5 < 3` | `False` |
| `>=` | มากกว่าหรือเท่ากับ | `5 >= 5` | `True` |
| `<=` | น้อยกว่าหรือเท่ากับ | `5 <= 3` | `False` |

**⚠️ ข้อผิดพลาดที่พบบ่อย:** `=` ใช้กำหนดค่า ส่วน `==` ใช้เปรียบเทียบ — สลับกันใช้ไม่ได้!


````python
x = 10
y = 20

print(x == y)   # False เพราะ 10 ไม่เท่ากับ 20
print(x != y)   # True เพราะ 10 ไม่เท่ากับ 20 จริง
print(x < y)    # True
print(x >= 10)  # True เพราะ 10 เท่ากับ 10
````

### 3.3 Logical Operators (ตัวดำเนินการตรรกะ)

ใช้รวมเงื่อนไขหลายๆ อันเข้าด้วยกัน

| Operator | ความหมาย | ตัวอย่าง |
|---|---|---|
| `and` | จริงทั้งคู่ จึงเป็นจริง | `True and False` → `False` |
| `or` | จริงอย่างน้อยหนึ่ง จึงเป็นจริง | `True or False` → `True` |
| `not` | กลับค่าตรงข้าม | `not True` → `False` |


````python
age = 20
has_id_card = True

# ต้องอายุ >= 18 AND มีบัตรประชาชน จึงเข้าได้
can_enter = (age >= 18) and (has_id_card == True)
print("เข้าได้ไหม:", can_enter)

# ตัวอย่าง or
is_weekend = False
is_holiday = True
can_rest = is_weekend or is_holiday
print("วันนี้พักได้ไหม:", can_rest)

# ตัวอย่าง not
is_raining = False
print("ไม่ฝนตกใช่ไหม:", not is_raining)
````

## 4. Type Conversion / Casting(การแปลงชนิดข้อมูล)

บางครั้งข้อมูลที่ได้มาไม่ใช่ชนิดที่เราต้องการ เช่น รับค่าจากผู้ใช้มาเป็น `str` เสมอ แต่เราต้องเอาไปคำนวณเป็นตัวเลข — ต้องแปลงชนิดข้อมูลก่อน

| ฟังก์ชัน | แปลงเป็น | ตัวอย่าง |
|---|---|---|
| `int()` | จำนวนเต็ม | `int("5")` → `5` |
| `float()` | ทศนิยม | `float("3.14")` → `3.14` |
| `str()` | ข้อความ | `str(5)` → `"5"` |
| `bool()` | จริง/เท็จ | `bool(1)` → `True` |


````python
str(False)
````

````python
bool([1])
````

````python
# str -> int / float
age_text = "20"
age_number = int(age_text)
print(age_number + 5)   # บวกได้แล้วเพราะเป็น int

price_text = "99.50"
price_number = float(price_text)
print(price_number * 2)

# int/float -> str (ใช้บ่อยตอน + ข้อความ)
score = 95
message = "คะแนนของคุณคือ " + str(score)
print(message)
````

**⚠️ ข้อผิดพลาดที่พบบ่อย:** ถ้า string ไม่ใช่ตัวเลข จะแปลงไม่ได้ และ Python จะแจ้ง error แบบ `ValueError`


````python
# ทดลองรันเซลล์นี้ดู จะเกิด error อะไร?
text = "สวัสดี"
# number = int(text)   # ลบ # ข้างหน้าแล้วรันดู จะเห็น ValueError

# วิธีป้องกัน: เช็คก่อนแปลงด้วย .isdigit()
text2 = "123"
if text2.isdigit():
    print(int(text2) + 1)
else:
    print("แปลงเป็นตัวเลขไม่ได้")
````

````python
# bool <-> int: True เท่ากับ 1, False เท่ากับ 0
print(int(True))    # 1
print(int(False))   # 0
print(bool(0))       # False
print(bool(1))       # True
print(bool(-99))      # True -- ตัวเลขที่ไม่ใช่ 0 ทั้งหมดเป็น True!
print(bool(""))      # False -- string ว่างเป็น False
print(bool("hello")) # True -- string ที่ไม่ว่างเป็น True
````

## 5. if / elif / else (การเขียนเงื่อนไข)

ใช้ตัดสินใจว่าโปรแกรมควรทำอะไร ขึ้นอยู่กับเงื่อนไขที่กำหนด

### โครงสร้างพื้นฐาน

```python
if เงื่อนไข1:
    # ทำสิ่งนี้ ถ้าเงื่อนไข1 เป็นจริง
elif เงื่อนไข2:
    # ทำสิ่งนี้ ถ้าเงื่อนไข1 ไม่จริง แต่เงื่อนไข2 เป็นจริง
else:
    # ทำสิ่งนี้ ถ้าไม่มีเงื่อนไขไหนเป็นจริงเลย
```

**⚠️ จุดสำคัญที่สุดของ Python:** ต้องมี **เครื่องหมาย `:`** ท้ายบรรทัดเงื่อนไข และ**บรรทัดถัดไปต้องเยื้อง (indent)** ด้วย Tab หรือเว้นวรรค 4 ครั้ง — Python ใช้การเยื้องบรรทัดแทนวงเล็บปีกกา `{}` แบบภาษาอื่น


````python
# if แบบพื้นฐาน
temperature = 37

if temperature > 30:
    print("วันนี้อากาศร้อนมาก")
````

````python
# if-else
score = 55

if score >= 50:
    print("ผ่าน")
else:
    print("ไม่ผ่าน")
````

````python
# if-elif-else: ใช้ตัดเกรดจากคะแนน
score = 49

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
elif score >= 50:
    grade = "D"
else:
    grade = "F"

print(f"คะแนน {score} ได้เกรด {grade}")
````

### เงื่อนไขซ้อนกัน (Nested if) และเงื่อนไขซับซ้อน

สามารถใส่ `if` ซ้อนใน `if` ได้ และรวมหลายเงื่อนไขด้วย `and` / `or`


````python
# nested if: เช็คสมัครสมาชิกแบบมีเงื่อนไขซ้อน
age = 16
has_parent_consent = False

if age >= 18:
    print("สมัครได้เลย")
else:
    if has_parent_consent:
        print("อายุไม่ถึง แต่มีผู้ปกครองยินยอม สมัครได้")
    else:
        print("สมัครไม่ได้ ต้องรอผู้ปกครองยินยอมก่อน")
````

````python
# รวมเงื่อนไขด้วย and / or ในบรรทัดเดียว (อ่านง่ายกว่า nested if)
age = 16
has_parent_consent = True

if age >= 18 or (age >= 13 and has_parent_consent):
    print("สมัครได้")
else:
    print("สมัครไม่ได้")
````

### ข้อ 4: if/elif/else — ตรวจ BMI
สูตรคำนวณ BMI คือ `น้ำหนัก(kg) / (ส่วนสูง(m) ** 2)`

จงเขียนโค้ดที่:
1. กำหนดตัวแปร `weight = 65` (kg) และ `height = 1.70` (m)
2. คำนวณค่า BMI
3. ใช้ if/elif/else แปลผลตามเกณฑ์นี้:
   - BMI < 18.5 → "น้ำหนักน้อย"
   - 18.5 ≤ BMI < 23 → "น้ำหนักปกติ"
   - 23 ≤ BMI < 25 → "น้ำหนักเกิน"
   - BMI ≥ 25 → "อ้วน"
4. print ค่า BMI (ทศนิยม 2 ตำแหน่ง) และผลแปลความหมาย

💡 Hint: ใช้ `round(ค่า, 2)` เพื่อปัดทศนิยม 2 ตำแหน่ง


````python
# เขียนคำตอบข้อ 4 ที่นี่
weight = 65
height = 1.70
bmi = weight / height**2
if bmi < 18.5:
    result = "น้ำหนักน้อย"
elif bmi < 23:
    result = "น้ำหนักปกติ"
elif bmi < 25:
    result = "น้ำหนักเกิน"
else:
    result = "อ้วน"
print(f"BMI = {round(bmi, 2)}")
print(f"ผลการแปลความหมาย: {result}")
````

### ข้อ 8: แปลงชนิดข้อมูลจากแบบสอบถาม

age_text = "21"

4. print ด้วย f-string ให้ได้ผลลัพธ์ตามรูปแบบนี้ (รายได้มีคอมม่าคั่นหลักพันและทศนิยม 2 ตำแหน่ง):

อายุ 21 ปี รายได้ 15,000.50 บาท/เดือน (ประมาณ 500.02 บาท/วัน) พึงพอใจ: True

### ข้อ 9: คำนวณส่วนลดร้านค้า (if/elif/else ร่วมกับ and/or)
ร้านค้าแห่งหนึ่งมีเงื่อนไขส่วนลดดังนี้:

| เงื่อนไข | ส่วนลด |
|---|---|
| เป็นสมาชิก **และ** ยอดซื้อตั้งแต่ 1,000 บาทขึ้นไป | 15% |
| เป็นสมาชิก แต่ยอดซื้อต่ำกว่า 1,000 บาท | 10% |
| ไม่เป็นสมาชิก แต่ยอดซื้อตั้งแต่ 2,000 บาทขึ้นไป | 5% |
| กรณีอื่นๆ | ไม่มีส่วนลด |

กำหนดตัวแปร `total = 1500` (บาท) และ `is_member = True` จงคำนวณส่วนลด แล้ว print ในรูปแบบ:

```
ยอดซื้อ 1,500.00 บาท | ส่วนลด 15% = 225.00 บาท | ยอดสุทธิ 1,275.00 บาท
```

ทดสอบให้ครบทั้ง 4 กรณี:

| กรณี | `total` | `is_member` |
|---|---|---|
| 1 | 1500 | `True` |
| 2 | 800 | `True` |
| 3 | 2500 | `False` |
| 4 | 1500 | `False` |

💡 Hint: ลำดับของเงื่อนไขใน if/elif มีผลต่อคำตอบ ลองคิดว่าควรเช็คเงื่อนไขไหนก่อน

````python
# เขียนคำตอบข้อ 9 ที่นี่
total = 1500
is_member = True
if is_member and total >= 1000:
    rate = 15
elif is_member:
    rate = 10
elif total >= 2000:
    rate = 5
else:
    rate = 0

discount = total * rate / 100
net = total - discount

print(f"ยอดซื้อ {total:,.2f} บาท | ส่วนลด {rate}% = {discount:,.2f} บาท | ยอดสุทธิ {net:,.2f} บาท")
````

### ข้อ 10: ตัดเกรดรายวิชาจากคะแนนหลายส่วน (Nested if)
รายวิชาหนึ่งคิดคะแนนเต็ม 100 จาก 3 ส่วน:
- คะแนนงาน (`assignment`) เต็ม 30 คะแนน ใช้ตามจริง
- คะแนนสอบกลางภาค (`midterm`) เต็ม 100 คะแนน คิดเป็น 30%
- คะแนนสอบปลายภาค (`final`) เต็ม 100 คะแนน คิดเป็น 40%

นอกจากนี้ ถ้าเวลาเข้าเรียน (`attendance`) **ต่ำกว่า 80%** จะได้เกรด F ทันที ไม่ว่าคะแนนรวมจะเป็นเท่าไร

เกณฑ์เกรด (กรณีเวลาเรียนครบ):

| คะแนนรวม | เกรด |
|---|---|
| ≥ 80 | A |
| 75 ถึงต่ำกว่า 80 | B+ |
| 70 ถึงต่ำกว่า 75 | B |
| 65 ถึงต่ำกว่า 70 | C+ |
| 60 ถึงต่ำกว่า 65 | C |
| 55 ถึงต่ำกว่า 60 | D+ |
| 50 ถึงต่ำกว่า 55 | D |
| < 50 | F |

จงคำนวณคะแนนรวม แล้ว print ในรูปแบบ:

```
คะแนนรวม 78.00 / 100 เกรด B+
```

กรณีเวลาเรียนไม่ถึง 80% ให้ print ว่า:

```
คะแนนรวม 89.50 / 100 เกรด F (เวลาเรียนไม่ถึง 80%)
```

ทดสอบให้ครบทั้ง 3 กรณี:

| กรณี | `assignment` | `midterm` | `final` | `attendance` |
|---|---|---|---|---|
| 1 | 25 | 70 | 80 | 90 |
| 2 | 28 | 85 | 90 | 75 |
| 3 | 15 | 50 | 45 | 100 |

💡 Hint: ใช้ if ชั้นนอกเช็คเวลาเรียนก่อน แล้วค่อยใช้ if/elif/else ชั้นในตัดเกรด

````python
# เขียนคำตอบข้อ 10 ที่นี่
assignment = 25
midterm = 70
final = 80
attendance = 90

total = assignment + (midterm * 0.30) + (final * 0.40)

if attendance < 80:
    grade = "F"
    print(f"คะแนนรวม {total:.2f} / 100 เกรด {grade} (เวลาเรียนไม่ถึง 80%)")
else:
    if total >= 80:
        grade = "A"
    elif total >= 75:
        grade = "B+"
    elif total >= 70:
        grade = "B"
    elif total >= 65:
        grade = "C+"
    elif total >= 60:
        grade = "C"
    elif total >= 55:
        grade = "D+"
    elif total >= 50:
        grade = "D"
    else:
        grade = "F"

    print(f"คะแนนรวม {total:.2f} / 100 เกรด {grade}")
````

### วิธีที่ 1: เซฟจาก Colab ขึ้น GitHub ตรงๆ (ง่ายที่สุด)

1. ใน Colab ไปที่เมนู **File → Save a copy in GitHub**
2. ครั้งแรกจะให้ login และ authorize Google Colab ให้เข้าถึง GitHub account ของคุณ
3. เลือก repository ปลายทาง (หรือสร้าง repo ใหม่ได้จากหน้านี้เลย)
4. ตั้งชื่อไฟล์และ commit message แล้วกด **OK**
5. ไฟล์ `.ipynb` จะถูก push ขึ้น GitHub repo ที่เลือกทันที

### วิธีที่ 2: เปิด Notebook จาก GitHub เข้า Colab

1. ไปที่ **colab.research.google.com**
2. เลือกแท็บ **GitHub**
3. พิมพ์ชื่อ repo หรือ URL ของ Notebook บน GitHub
4. เลือกไฟล์ `.ipynb` ที่ต้องการเปิด

### วิธีที่ 3: ใช้ Git ผ่าน Terminal (สำหรับคนที่อยากฝึกใช้ Git จริงจัง)

```bash
git clone https://github.com/<username>/<repo-name>.git
cd <repo-name>
git add notebook.ipynb
git commit -m "เพิ่ม notebook คาบที่ 2"
git push origin main
```

> 💡 **Tip:** การเก็บ Notebook ทุกคาบไว้ใน GitHub repo เดียว (เช่น `SC612104-coursework`) จะกลายเป็น portfolio ที่โชว์ progress การเรียนของคุณได้ดีมาก ลองทำเป็นกิจวัตรหลังจบทุกคาบดู

สอนเสร็จแล้วจะให้ส่งด้วย!
